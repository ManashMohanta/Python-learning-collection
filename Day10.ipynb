```python
# create a parametery no. to find to check it is prime or not
def check_prime(num):
    count=0
    for i in range(1, num+1):
        if num%i==0:
            count+=1
    if count==2:
            print(" is prime")
    else:
           print("is not prime")
check_prime(45)
```

```python
# multiple key with argument
def print_data(**data): # dict argument store multiple value
    print("data is->", data)
    for i, j in data.items():
        print(i,"->",j)
```

```python
 # local variable only present inside a block ,we can use same value out side of that block .And  if we defined local variable with same name with diff block then these two will be d
# global variable present for hole program
def factors(num):
    fact=[]                  
    for i in range(1, num+1):
      if num%i==0:
        fact.append(i)
    return fact 
def its_prime(number):
    if len (factors(number))>2:
        print("its not prime")
    else:
        print("its prime")
                                    
```

```python
its_prime(5)
```

```python
## create a function to check user having valid username or not
# first letter should be an alphabet
# no special character should inside it expect"_"
# length should be more than 5 and less than 12
def valid_username(username):
    if (username.islower() and 5<= len(username)<=12 and all (ch.islower() or ch.isdigit() or ch=='_'for ch in username) and username.isalpha()):
        print("username is correct")
    else:
        print("please enter a valid username")
```
