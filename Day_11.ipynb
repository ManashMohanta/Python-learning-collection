```python
# prime number 
def is_prime(num):
    for i in range (2,num):
        if num%i==0:
            return False
    return True
def check_prime(num):
    if is_prime(num):
        print("prime number")
    else:
        print("not a prime number")
check_prime(67)
type(is_prime(67)) # type check kar rhe ha isliye bool dikaya and if without type then only true ata
```

```python
# sum of number from 1 
def range_sum(num):
    sum=0
    for i in range(1, num+1):
        sum=sum+i
    return sum
range_sum(78)

def sum_of_num(num):
    if num==1:
        return 1
    else:
        return num+sum_of_num(num-1)
sum_of_num(5)
```

```python
# create a recurtion fuction to print factorial of a no.
def factorial_of_num(num):
    if num==1:
        return 1
    else:
        return num*factorial_of_num(num-1)
factorial_of_num(5)
```

```python
# write a program for sum of digits
def sum_of_digit(num):
    if num==0:
        return 0
    else:
        return num%10+sum_of_digit(num//10)
sum_of_digit(537)
```
