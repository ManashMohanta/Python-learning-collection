```python
# list comprehension 
to_100=[x for x in range(1,101)] # braket ke under 1st x ha bo sab iteration complete honeke baad store kar dega 1st x mai
print(to_100) #  fir x ka assign value to_100 mai aajyega fir print karna hota ha usko
```

```python
# list comprehension 
to_100=[x for x in range(1,101) if x%2==0]
print(to_100)
```

```python
#  list comprehension
to_50=[x for x in range(1,51) if x%2!=0]
print("the odd values are",to_50)
```

```python
#  list comprehension with lambda fun()
square_upto_10=[list(map(lambda x:x**2, range (1,11)))]
square_upto_10
```

```python
#  list comprehension with lambda fun()
square_upto_10=[x**2 for x in range (1,11)]
square_upto_10
```

```python
# lambda in list comprehension 
a=[45,67,9,89,34, 90,56]
g_60= list(filter(lambda num: True if num>60 else False , a))
g_60
```

```python
b=[54,56,78,90,67,84,478,8]
gt_60=[(lambda x:"high" if x>60 else "low")(x) for x in b]
gt_60
```

```python
b=[54,56,78,90,67,84,478,8]
gt_60=[(lambda x:x if x>60 else 0) (x) for x in b]
gt_60
```

```python
class cookie:
    bakery_name=None
    def flavour_of_it(self,f):
        self.flavour=f
    def print_cookie(self):
        print(f"this is delicious {self.flavour} cookie")
c1=cookie()
c1.bakery_name="cookie bite"
c1.flavour_of_it("chocolate")
print(f"welcome to {c1.bakery_name}")
c1.print_cookie()
```
