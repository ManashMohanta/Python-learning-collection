```python
# create a class circle that calculate the area and perimeter where the radius is  a class attribute
class circle:
    pi=3.14
    def fun(self,r):
        print(f"Area of circle:{(self.pi*r**2)}")
        print(f"perimeter of circle:{2*self.pi*r}")
obj3=circle()
obj3.fun(12)
```

```python
# create a class named rectangle with attribute length and width, abd method to calculate area of perimeter 
class Rectangle:
    l=12
    w=12
    def abd(self):
        print(f"Area of Rectrangle :{self.l*self.w}")
        print(f"perimeter of Rectrangle:{2*self.l+self.w}")
obj4=Rectangle()
obj4.abd()
```
