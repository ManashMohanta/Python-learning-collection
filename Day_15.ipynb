```python
# Develop a class car
class car:
    company="BMW" # class attribute hai company= Mahindra
    def __init__(self,brand, model,fuel,km):
        self.brand=brand # object attribute is can be change every object
        self.model=model
        self.fuel_percent=fuel
        self.km_driven=km
    def display(self):
        print("The company name is ", self.company)
        print("The brand is ", self.brand)
        print("The model name is ", self.model)
        print("The fuel percentage is ", self.fuel_percent,"%")
        print("The total km driven is ", self.km_driven)
    def fill(self,amount):
        self.fuel_percent+=amount
        
        print("your updated fuel percent is ",self.fuel_percent,"%")
        
    def drive(self,amount):
        self.km_driven+=amount
        self.fuel_percent-=10
        
        print(f"your fuel remaining {self.fuel_percent} % And you drove {self.km_driven} km")
```

```python
c1=car("BMW M series","sendan",80,3000)
c1.display()
c1.fill(12)
c1.drive(100)
```

```python
c2=car("BMW  i series","SUV",50,2000)
c2.display()
c2.fill(12)
c2.drive(100)
```

```python
# Single Inhertance
class A:
    name="hari"
    def print_A(self):
        print("my name is class A")
class B(A): # B is inherit the property of A
    def print_B(self):
        print("my name is class B")
ob=B()
ob.name
ob.print_A()
```

```python
# multiple inheritance
class A:
    name="hari"
    def print_A(self):
        print("my name is class A")
class B:
    def print_B(self):
        print("my name is class B")
class C(A,B):
    def print_C(self):
        print("my name is class C")
ob=C()
ob.name
ob.print_C()
```

```python
# Herirachical inheritance
class A:
    name="hari"
    def print_A(self):
        print("my name is class A")
class B(A): 
    def print_B(self):
        print("my name is class B")
class C(A):
    def print_C(self):
        print("my name is class C")
ob=A()
ob.name
ob.print_A()
```

```python
# Multilevel inheritance
class A:
    name="hari"
    def print_A(self):
        print("my name is class A")
class B(A): 
    def print_B(self):
        print("my name is class B")
class C(B):
    def print_C(self):
        print("my name is class C")
ob=A()
ob.name
ob.print_A()
```
