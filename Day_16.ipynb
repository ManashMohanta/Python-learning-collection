```python
class Calc:
    def __init__(self,a,b):
        self.first=a
        self.second=b
    def add(self):
        print(f"{self.first}+{self.second}={self.first+self.second}")
    def sub(self):
        print(f"{self.first}-{self.second}={self.first-self.second}")
    def mul(self):
        print(f"{self.first}*{self.second}={self.first*self.second}")
    def div(self):
        print(f"{self.first}/{self.second}={self.first/self.second}")
    def modulus(self):
        print(f"{self.first}%{self.second}={self.first%self.second}")
nums1=Calc(45,56)
nums1.add()
nums1.sub()
nums1.mul()
nums1.div()
nums1.modulus()
```

```python
class EduSys:

    def AddCourse(self, value):
        self.course_name = value
        return self.course_name

    def AddTeacher(self, value):
        self.teacher_name = value
        return self.teacher_name

    def AddStudent(self, value):
        self.student_name = value
        return self.student_name


class Student(EduSys):

    def __init__(self, name, course, teacher):
        EduSys.AddStudent(self, name)
        EduSys.AddCourse(self, course)
        EduSys.AddTeacher(self, teacher)

    def student_details(self):
        print(f"The student name is => {self.student_name}")
        print(f"Teacher asign for student is => {self.teacher_name}")
        print(f"Currently student in {self.course_name} course")


s1 = Student(name="rohan", course="Python", teacher="Ram Lal")
s1.student_details()
```
