```python
# create a program to check the year i leap year or not?(nested if)
year=float(input("enter year"))
if (year%100!=0):
    if (year%4==0):
        print("it is a leap year")
    else:
        print(" it is not leap year")
else:
    if(year%400==0):     
        print (" it is a leap year")
    else :
        print(" it is not a leap year")
```
