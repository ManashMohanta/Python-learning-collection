```python
# create a program to print sum of digits of a no.
num= int(input("Enter a number"))
sum=0
for i in range(len(str(num))):
    digit=num%10
    sum+=digit
    num//=10

print("you sum is->", sum)
```

```python
# create a progam to reverse a number (using for loop)
num = int(input("Enter a number->"))
rev=0
for i in range(len(str(num))):
    digit= num%10
    rev= (rev*10)+digit#(rev*10) wor as a exponential form 10 power 0
    num//=10
print("reverse of number is", rev)
```
