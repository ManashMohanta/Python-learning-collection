```python
#Write a Python program that takes an integer as input from the user and finds the smallest digit present in the number.
#Enter a number: 583921
#Smallest digit: 1
num=int(input("enter a no:"))
smallest=9
while num!=0:
    digit=num%10
    if smallest>digit:
       smallest=digit
    num//=10
print("smallest digit is = ",smallest)
```
